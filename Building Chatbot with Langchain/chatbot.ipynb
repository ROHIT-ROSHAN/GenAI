```python
import os
from dotenv import load_dotenv
load_dotenv()
## Loading all the env variables

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_openai import ChatOpenAI
from langchain_groq import ChatGroq
model=ChatGroq(model="openai/gpt-oss-120b",groq_api_key=groq_api_key )
```

```python
from langchain_core.messages import HumanMessage
model.invoke(
    [
        HumanMessage(content="Hi My name is Rohit, I am a chief AI Engineer"),

    ]
)
```

```python
from langchain_core.messages import AIMessage
model.invoke(
    [
        HumanMessage(content="Hi My name is Rohit, I am a chief AI Engineer"),
        AIMessage(content="Hello Rohit! 👋\n\nGreat to meet a chief AI Engineer. How can I help you today? Whether you’re looking for technical insights, brainstorming ideas, troubleshooting a model, or anything else—just let me know!"),
        HumanMessage(content="Hey what is my name and what do I do")
    ]
)
```

```python
### Message History
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory


store  = {}
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]= ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model , get_session_history)
```

```python
config = {"configurable":{"session_id":"chat1"}}
```

```python
response=with_message_history.invoke(
    [
        HumanMessage(content="Hi My name is Rohit, I am a chief AI Engineer"),
    ],
    config=config
)
```

```python
response.content
```

```python
response=with_message_history.invoke(
    [
        HumanMessage(content="Hi What is my name, and what do I do?"),
    ],
    config=config
)
```

```python
response.content
```
